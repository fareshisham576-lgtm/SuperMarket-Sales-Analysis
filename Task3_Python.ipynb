## *Data Cleaning*

```python
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip()
```

```python
df.info()
```

```python
df['Branch'].unique()
```

```python
df['Branch'] = df['Branch'].str.strip().replace({'giza':'Giza' , 'cairo':'Cairo'})
```

```python
df['Branch'].unique()
```

```python
df['City'].unique()
```

```python
df['Customer type'].unique()
```

```python
df['Gender'].unique()
```

```python
df['Gender'] = df['Gender'].replace({'F':'Female' , 'M':'Male'})
```

```python
df['Gender'].unique()
```

```python
df['Product line'].unique()
```

```python
df['Unit price'].astype('float64')
```

```python
df['Unit price'] = df['cogs']/df['Quantity']
```

```python
df.info()
```

```python
df['Sales'] = df['Quantity'] * df['Unit price'] + df['Tax 5%']
```

```python
df['Sales'].astype("float64")
```

```python
df.info()
```

```python
df.dropna(inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df["Day"] = df['Date'].dt.day_name()
df["Month"] = df['Date'].dt.month_name()
df["Year"] = df['Date'].dt.year
```

```python
df.head()
```

```python
df.info()
```

```python
df['Time'] = pd.to_datetime(df['Time']).dt.strftime('%H:%M:%S')
```

```python
df.head()
```

```python
df['Payment'].unique()
```

```python
df['Payment'] = df['Payment'].replace({
    'Ewallet':'E-Wallet',
    'cash':'Cash',
    'credit card':'Credit card'
})
```

```python
df['Payment'].unique()
```

```python
# Cogs = Quantity * Unit Price so i will change name to "Total Price" , Sales to Tatal Salary   ,    Tax 5% to Tax(TS - TP)
df.rename(columns={'cogs':'Total Price' , 'Sales':'Total Salary' , 'Tax 5%':'Tax(TS - TP)'}, inplace=True)
```

```python
df.head()
```

```python
# gross margin percentage is the same number , we can drop column
# gross income = Tax 5% ,drop
# drop Invoice id because we don't benefit from it
# drop City because Branch is enough
df.drop(columns=['gross margin percentage', 'gross income', 'Invoice ID' , 'City'], inplace=True)
```

```python
df.head()
```

```python
def satisfaction(rating):
  if rating >= 7:
    return "Satisfied"
  else:
    return "Not Satisfied"
```

```python
df["Satisfaction"] = df["Rating"].apply(satisfaction)
```

```python
df.sort_values(by='Total Salary', ascending=False, inplace=True)
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
df.to_csv('Cleaned_SuperMarket.csv', index=False)
```

```python
# 1)Provide the total revenue generated by the supermarket.
print(f'Total Reveue: {df['Total Salary'].sum().round(1)} $')
```

```python
# 2) Provide the total revenue by city and identify the city with the highest revenue.
df.groupby('Branch')['Total Salary'].sum().sort_values(ascending=False).round(1)
```

```python
print(f"The City that have the Highest Revenue: {df.groupby('Branch')['Total Salary'].sum().idxmax()}")
```

```python
# 3) Provide the total profit by branch and identify the most profitable branch.
df.groupby('Branch')['Total Price'].sum().sort_values(ascending=False).round(1)
```

```python
print(f"The City that have the Highest Profit: {df.groupby('Branch')['Total Price'].sum().idxmax()}")
```

```python
# 4) Provide the total revenue and profit by product category and identify the top-performing category.
df.groupby('Product line')[['Total Salary', 'Total Price']].sum().sort_values(by='Total Salary', ascending=False).round(1)
```

```python
print(f"The Product Category that have the Highest Revenue: {df.groupby('Product line')['Total Salary'].sum().idxmax()}")
```

```python
fig1 = px.bar(df, x='Product line', y='Total Salary', color='Branch', title='Total Revenue by Product Category')
fig1.show()
```

```python
# 5) Provide the total spending by customer type and identify which customer type spends more.
df.groupby('Customer type')['Total Salary'].sum().sort_values(ascending=False).round(1)
```

```python
print(f"The Customer Type that have the Highest Spending: {df.groupby('Customer type')['Total Salary'].sum().idxmax()}")
```

```python
# 6) Provide the number of transactions by payment method and identify the most popular payment method.
df['Payment'].value_counts()
```

```python
print(f"The Payment Method that have the Highest Transactions: {df['Payment'].value_counts().idxmax()}")
```

```python
fig2 = px.pie(df, names='Payment', title='Number of Transactions by Payment Method')
fig2.show()
```

```python
# 7) Provide the average transaction value for the supermarket.
print(f'The Average Transaction Value: {df["Total Salary"].mean().round(1)} $')
```

```python
# 8) Provide the customer satisfaction percentage by branch and identify the branch with the highest satisfaction.
df.groupby('Branch')['Satisfaction'].value_counts(normalize=True).mul(100).round(1).unstack()
```

```python
print(f"The Branch that have the Highest Satisfaction: {df.groupby('Branch')['Satisfaction'].value_counts(normalize=True).mul(100).idxmax()}")
```

```python
# 9) Provide the total revenue by day and by month and identify the day and month with the highest sales.
df.groupby('Day')['Total Salary'].sum().sort_values(ascending=False).round(1)
```
